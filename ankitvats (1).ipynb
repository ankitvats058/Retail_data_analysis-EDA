```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.countplot(
    data=df,
    x="payment_method",
    ax=axes[0],
    palette="Set2",
    hue="payment_method",
    legend=False,
)
axes[0].set_title("Payment Method Distribution")

sns.countplot(
    data=df,
    x="return_status",
    ax=axes[1],
    palette="Set1",
    hue="return_status",
    legend=False,
)
axes[1].set_title("Return Status Distribution")

sns.countplot(
    data=df,
    x="customer_segment",
    ax=axes[2],
    palette="Set3",
    hue="customer_segment",
    legend=False,
)
axes[2].set_title("Customer Segment Distribution")

plt.tight_layout()
plt.show()
```

```python
Q1 = df["final_price"].quantile(0.25)
Q3 = df["final_price"].quantile(0.75)
IQR = Q3-Q1
lower_bound = Q1-1.5*IQR
upper_bond = Q3+1.5*IQR
iqr_outliers = df[(df["final_price"]< lower_bound) | (df["final_price"]>upper_bond)]
print(f"IQR Method: Total outliers found in final_price = {len(iqr_outliers)}")
                                                      
```

```python
from scipy import stats

z_scores = stats.zscore(df["final_price"])
z_outliers = df[abs(z_scores) > 3]

print(f"Z-score Method: Total outliers found (Z > 3) = {len(z_outliers)}")
```

```python
plt.figure(figsize=(10, 5))
sns.barplot(
    data=df,
    x="product_category",
    y=(df["return_status"] == "Returned").astype(int),
    errorbar=None,
    palette="viridis",
    hue="product_category",
    legend=False,
)
plt.title("Return Rate by Product Category")
plt.ylabel("Return Rate (Percentage)")
plt.xticks(rotation=45)
plt.show()
```
